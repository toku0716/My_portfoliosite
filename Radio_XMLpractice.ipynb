```python
import requests
import xml.etree.ElementTree as ET
```

```python
url = "https://radiko.jp/v3/program/station/date/20261011/QRR.xml"
response = requests.get(url,timeout=15)
response.raise_for_status()

root = ET.fromstring(response.content)

for prog in root.iter("prog"):
    title = prog.findtext("title",default="")
    pfm = prog.findtext("pfm",default="")
    pfm = prog.findtext("time",default="")
    start_time = prog.get("ft","")
    end_time = prog.get("to","")

    if start_time:
      start_time=start_time[8:10]+":"+start_time[10:12]

    if end_time:
        end_time=end_time[8:10]+":"+end_time[10:12]
    print("放送時間：",start_time,"~",end_time)

    print("番組表",title)
    print("出演者",pfm)
```
